```python

# Aura AI — Step 3
# Create synthetic cybersecurity data

import numpy as np
import pandas as pd

np.random.seed(42)

normal_count = 1000
anomaly_count = 50

# -----------------------------
# Normal activity
# -----------------------------

normal_data = pd.DataFrame({
    "login_frequency": np.random.normal(5, 1.5, normal_count),
    "failed_logins": np.random.poisson(1, normal_count),
    "new_device": np.random.binomial(1, 0.05, normal_count),
    "api_requests": np.random.normal(100, 20, normal_count),
    "unusual_hour": np.random.binomial(1, 0.05, normal_count),
    "data_transfer_mb": np.random.normal(50, 15, normal_count)
})

# -----------------------------
# Abnormal activity
# -----------------------------

anomaly_data = pd.DataFrame({
    "login_frequency": np.random.normal(20, 5, anomaly_count),
    "failed_logins": np.random.poisson(10, anomaly_count),
    "new_device": np.random.binomial(1, 0.8, anomaly_count),
    "api_requests": np.random.normal(500, 100, anomaly_count),
    "unusual_hour": np.random.binomial(1, 0.8, anomaly_count),
    "data_transfer_mb": np.random.normal(500, 100, anomaly_count)
})

# Combine datasets
data = pd.concat(
    [normal_data, anomaly_data],
    ignore_index=True
)

# Prevent impossible negative values
numeric_columns = [
    "login_frequency",
    "failed_logins",
    "api_requests",
    "data_transfer_mb"
]

for column in numeric_columns:
    data[column] = data[column].clip(lower=0)

print("Aura AI cybersecurity dataset created.")
print()
print("Total events:", len(data))
print("Normal events:", normal_count)
print("Synthetic anomalous events:", anomaly_count)

print("\nFirst 10 records:")
display(data.head(10))
```

```python

# Aura AI — Step 4
# Train Isolation Forest anomaly-detection model

from sklearn.ensemble import IsolationForest

# Features used by the ML model
features = [
    "login_frequency",
    "failed_logins",
    "new_device",
    "api_requests",
    "unusual_hour",
    "data_transfer_mb"
]

X = data[features]

# Create the Isolation Forest model
model = IsolationForest(
    n_estimators=200,
    contamination=50 / len(data),
    random_state=42
)

# Train the model
model.fit(X)

print("Aura AI Isolation Forest model trained successfully.")
print("Number of trees:", model.n_estimators)
print("Training samples:", len(X))
print("Features used:", len(features))
```

```python

# Aura AI — Step 5
# Detect anomalous security events

# Predict:
#   1  = normal
#  -1  = anomaly
predictions = model.predict(X)

# Calculate anomaly scores
scores = model.decision_function(X)

# Add results to the dataset
results = data.copy()

results["prediction"] = predictions
results["anomaly_score"] = scores

# Convert prediction into a readable label
results["status"] = results["prediction"].map({
    1: "NORMAL",
    -1: "ANOMALY"
})

# Count results
normal_detected = (predictions == 1).sum()
anomalies_detected = (predictions == -1).sum()

print("Aura AI anomaly detection completed.")
print()
print("Normal events detected:", normal_detected)
print("Anomalies detected:", anomalies_detected)

print("\nDetected anomalies:")
display(
    results[results["status"] == "ANOMALY"]
    .sort_values("anomaly_score")
    .head(20)
)
```

```python

# Aura AI — Step 6
# Convert anomaly scores into security risk levels

def calculate_risk(score):
    if score < -0.08:
        return "CRITICAL"
    elif score < -0.04:
        return "HIGH"
    elif score < 0:
        return "MEDIUM"
    else:
        return "LOW"


results["risk_level"] = results["anomaly_score"].apply(calculate_risk)

# Show the highest-risk events
high_risk_events = results.sort_values(
    "anomaly_score"
).head(20)

print("Aura AI Risk Engine completed.")
print("\nHighest-risk events:")

display(
    high_risk_events[
        features + ["anomaly_score", "status", "risk_level"]
    ]
)
```

```python

# Aura AI — Step 7
# Explain suspicious security events

def explain_event(row):
    reasons = []

    if row["login_frequency"] > 12:
        reasons.append("Unusually high login frequency")

    if row["failed_logins"] >= 5:
        reasons.append("High number of failed login attempts")

    if row["new_device"] == 1:
        reasons.append("Activity from a new device")

    if row["api_requests"] > 250:
        reasons.append("Unusually high API request volume")

    if row["unusual_hour"] == 1:
        reasons.append("Activity occurred during an unusual hour")

    if row["data_transfer_mb"] > 250:
        reasons.append("Unusually large data transfer")

    if not reasons:
        reasons.append("No individual threshold exceeded; anomaly detected from combined behavior")

    return reasons


# Select the highest-risk event
event = results.sort_values("anomaly_score").iloc[0]

reasons = explain_event(event)

print("AURA AI SECURITY ANALYSIS")
print("=" * 35)
print("Event ID:", event.name)
print("Status:", event["status"])
print("Risk Level:", event["risk_level"])
print("Anomaly Score:", round(event["anomaly_score"], 4))

print("\nDetected signals:")
for reason in reasons:
    print("•", reason)

print("\nRecommended action:")
if event["risk_level"] == "CRITICAL":
    print("Immediately investigate the event and temporarily restrict the affected session if appropriate.")
elif event["risk_level"] == "HIGH":
    print("Investigate the event and increase monitoring.")
else:
    print("Continue monitoring the activity.")
```

```python

# Aura AI — Step 9
# Create reusable anomaly detector

anomaly_detector_code = r'''
import joblib
import pandas as pd
import os

MODEL_PATH = os.path.join(
    os.path.dirname(__file__),
    "saved_model",
    "isolation_forest.joblib"
)

FEATURES = [
    "login_frequency",
    "failed_logins",
    "new_device",
    "api_requests",
    "unusual_hour",
    "data_transfer_mb"
]


class AuraAnomalyDetector:

    def __init__(self):
        self.model = joblib.load(MODEL_PATH)

    def predict(self, event):
        """
        Analyze one security event.

        Returns:
            dict containing anomaly status and score.
        """

        # Convert input event to DataFrame
        df = pd.DataFrame([event])

        # Ensure the required features exist
        missing = [feature for feature in FEATURES if feature not in df]

        if missing:
            raise ValueError(
                f"Missing required features: {missing}"
            )

        X = df[FEATURES]

        # ML prediction
        prediction = self.model.predict(X)[0]

        # Anomaly score
        score = float(
            self.model.decision_function(X)[0]
        )

        status = (
            "ANOMALY"
            if prediction == -1
            else "NORMAL"
        )

        return {
            "status": status,
            "anomaly_score": round(score, 6)
        }
'''

with open("aura_ml/anomaly_detector.py", "w") as f:
    f.write(anomaly_detector_code)

print("Created:")
print("aura_ml/anomaly_detector.py")
```

```python

# Aura AI — Step 9
# Colab-compatible anomaly detector

import joblib
import pandas as pd
import os


# Model location in Google Colab
MODEL_PATH = "aura_ml/saved_model/isolation_forest.joblib"


FEATURES = [
    "login_frequency",
    "failed_logins",
    "new_device",
    "api_requests",
    "unusual_hour",
    "data_transfer_mb"
]


class AuraAnomalyDetector:

    def __init__(self):
        self.model = joblib.load(MODEL_PATH)

    def calculate_risk(self, score):

        if score < -0.08:
            return "CRITICAL"

        elif score < -0.04:
            return "HIGH"

        elif score < 0:
            return "MEDIUM"

        else:
            return "LOW"

    def explain_event(self, event):

        reasons = []

        if event["login_frequency"] > 12:
            reasons.append(
                "Unusually high login frequency"
            )

        if event["failed_logins"] >= 5:
            reasons.append(
                "High number of failed login attempts"
            )

        if event["new_device"] == 1:
            reasons.append(
                "Activity from a new device"
            )

        if event["api_requests"] > 250:
            reasons.append(
                "Unusually high API request volume"
            )

        if event["unusual_hour"] == 1:
            reasons.append(
                "Activity occurred during an unusual hour"
            )

        if event["data_transfer_mb"] > 250:
            reasons.append(
                "Unusually large data transfer"
            )

        if not reasons:
            reasons.append(
                "Anomaly detected from combined behavior"
            )

        return reasons

    def predict(self, event):

        df = pd.DataFrame([event])

        missing = [
            feature
            for feature in FEATURES
            if feature not in df.columns
        ]

        if missing:
            raise ValueError(
                f"Missing required features: {missing}"
            )

        X = df[FEATURES]

        prediction = self.model.predict(X)[0]

        score = float(
            self.model.decision_function(X)[0]
        )

        status = (
            "ANOMALY"
            if prediction == -1
            else "NORMAL"
        )

        risk_level = self.calculate_risk(score)

        reasons = self.explain_event(event)

        return {
            "status": status,
            "risk_level": risk_level,
            "anomaly_score": round(score, 6),
            "reasons": reasons
        }


print("Aura Anomaly Detector loaded successfully.")
```
